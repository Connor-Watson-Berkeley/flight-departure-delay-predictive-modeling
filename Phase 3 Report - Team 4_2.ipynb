# Abstract

Flight delays cost airlines billions annually through crew scheduling disruptions, gate misallocations, and passenger compensation. This multi-phase project predicts domestic flight departure delays in minutes two hours before scheduled departure, enabling proactive airline operations and improved passenger experience. Using the U.S. Department of Transportation On-Time Performance dataset (70+ million flights from 2015–2021), integrated with NOAA weather observations and airport metadata, we identified key delay patterns: while 80% of flights operate within 15 minutes of schedule, delays are highly variable (mean 10 minutes, σ = 38) with elevated occurrences on Mondays, Sundays, and peak summer months.

We define delay as the difference in minutes between the scheduled departure time and the actual departure time. A flight is considered "delayed" if it is late by more than 15 minutes. It is considered severely delayed if it is late by more than 60 minutes. In Phases 1-2, we framed this as a regression problem predicting continuous delay times, evaluated using RMSE for interpretability and domain-specific metrics (On-Time Performance Prediction Accuracy and Severe Delay Detection Rate) for operational relevance. Our baseline linear regression model, built with a 6-stage feature engineering pipeline using expanding window cross-validation, achieved 35.96 minutes RMSE on the 12-month dataset and 47.92 minutes on the 60-month dataset, establishing clear performance benchmarks for subsequent model comparisons.

In Phase 3, we expanded our approach through sophisticated feature engineering. We developed four engineered feature families: (1) Flight Lineage features capturing cascading delay effects through aircraft rotations, (2) Graph-based PageRank features measuring airport network centrality, (3) Meta-models predicting previous flight durations (air time R² = 0.958, turnover time R² = 0.785, rotation time R² = 0.895) to avoid data leakage, and (4) comprehensive weather integration. Implementing Gradient Boosted Trees, XGBoost, and distributed PyTorch deep neural networks across GPU clusters, our final XGBoost model with Graph, Lineage, and Meta-Model features on the complete OTPW dataset achieved 37.76 minutes RMSE (12.5% improvement), 12.79 minutes MAE (23.1% improvement), 87.47% OTPA (+5.4%), and 62.94% SDDR F1 (+23.0% improvement), demonstrating that engineered features combined with data completeness are critical for capturing cascading delay patterns and enabling actionable predictions for airline decision-makers.

# Key Terms

- **3M, 12M, 60M**: Dataset sizes in months (M) of flight data. All datasets start on 2015-01-01 and contain 3, 12, or 60 months of data respectively.

- **Custom and OTPW**: Two dataset sources with equivalent structure but different data completeness. **Custom** refers to the "Custom Join" dataset created via our data pipeline, which experienced data loss from an errant join condition. **OTPW** refers to the pre-joined dataset provided, which maintains complete data without loss.

- **Cross-Validation Set and Test Set**: All datasets (3M, 12M, 60M) are split into temporal folds using a rolling window approach. The **cross-validation set** refers to the average results across 3-fold rolling window cross-validation (Folds 1-3), where each fold uses non-overlapping training and validation periods. The **test set** refers to a single model trained on all data from Folds 1-3 and evaluated on Fold 4 (the final temporal period). For the 60M dataset, models are trained on 2015-2018 data and evaluated on the 2019 test set.

## Data Pipeline & Custom Joins

We designed and implemented a custom data pipeline to robustly combine disparate data sources. A critical requirement was connecting flight and weather data both **spatially** (mapping airports to weather stations) and **temporally** (aligning flight times with weather observations). Additionally, we standardized timestamps, converting local flight times to UTC to match the weather data format.

The pipeline was executed in a series of optimized joins, described as follows:

#### 1. Flight–Airport Integration (IATA/ICAO Joins + Timezone Normalization)
The first stage enriches raw flight records with metadata from origin and destination airports.

- **Airport Metadata Construction:** We synthesized an airport master table by joining airport code datasets with timezone datasets. This was performed using both **IATA and ICAO identifiers**, ensuring maximum coverage for both major hubs and smaller regional airports.
  
- **Flight Enrichment & Optimization:** Cleaned airport attributes, including latitude, longitude, ISO country, and timezone, were joined to the flights dataset for both the origin and destination.  
  *Optimization:* We utilized a **broadcast join** hint for the airport metadata. Since the airport lookup table is small relative to the flight data, broadcasting it to all worker nodes minimized network shuffling and significantly improved performance.

- **UTC Time Construction:** Flight schedules are provided in local time (CRS departure time). Using the joined **origin airport timezone**, we derived the true UTC timestamp for every flight (`sched_depart_date_time_UTC`). 
  Critically, we pre-computed a **"Two-Hours-Prior"** timestamp (`two_hours_prior_depart_UTC`) to serve as the anchor for joining predictive weather data (simulating the information available 2 hours before takeoff).

#### 2. Spatial Mapping: Linking Airports to Nearest NOAA Weather Station
Because airports and weather stations are distinct entities, we mapped each origin airport to its nearest NOAA station using a two-stage spatial filter.

- **Bounding Box Pre-Filter:** To avoid a computationally expensive cross-join, we first filtered candidate stations using a **0.5-degree latitude/longitude bounding box** around each airport.

- **Precise Distance Calculation:** On the filtered candidates, we calculated the precise **Haversine (great-circle) distance**, filtering for stations within a strict **50 km** radius.

- **Nearest-Station Selection:** We employed a window function to rank candidate stations by distance and selected the nearest single station (`rank = 1`) for every airport. These mappings were cached to ensure consistency and speed in subsequent steps.

#### 3. Temporal Integration: Time-Window Weather Matching (Bucketed Join + Ranking)

<p align="center">
<img src='https://github.com/Connor-Watson-Berkeley/flight-departure-delay-predictive-modeling/raw/main/report-images/Sid-time-weather-match.png' width="60%"
    style="border:3px solid black;">
</p>

The most complex operation was aligning 60 million flights with high-frequency weather data. A standard join would be computationally prohibitive due to data skew. We implemented a **bucketed co-partitioning strategy** to efficiently join these datasets while ensuring no look-ahead bias.

**The Logic:**
1.  **Time Bucketing:** We assigned both flights (based on their 2-hour-prior timestamp) and weather records to **30-minute Unix-time buckets**.
2.  **Bucket Shifting for Continuity:** To ensure a flight at the beginning of a bucket could "see" weather from the immediate past, we duplicated weather records and shifted them forward by one bucket interval. This ensured that for any given partition, the join would have access to weather from both the current and the previous 30-minute window.
3.  **Co-Partitioned Join:** We repartitioned both datasets by `(Station_ID, Bucket_ID)`. This aligned the data physically across the cluster, allowing for a highly efficient local join without extensive shuffling.
4.  **Filtering & Ranking:**
    * **Filter:** We strictly kept weather records where `weather_timestamp <= flight_2hr_prior_timestamp`.
    * **Rank:** We calculated the time difference (`flight_ts - weather_ts`) and used a window function to select the single weather record closest in time to the flight's target timestamp.
5.  **Left Join Preservation:** The operation was structured as a Left Join, ensuring that flights without matching weather data were retained (with null weather features) rather than dropped.

### Engineered Features

In this section we will discuss the motivation and methodology behind our feature engineering strategy and their performance impact on our top performing model class, XGBoost. See "Engineered Features List" in the appendix for more details.

#### Motivation: The Fundamental Equation

$$\text{Delay} = \text{Available Time} - \text{Air Time} - \text{Turnover Time}$$

Where:
- **Available Time**: Time between previous flight's departure and current flight's scheduled departure
- **Air Time**: Actual flight duration (wheels-off to wheels-on)
- **Turnover Time**: Ground operations time (previous arrival to current departure)

When **Available Time < (Air Time + Turnover Time)**, the current flight departs late.

#### Delay Propagation Mechanisms

1. **Late Arrivals Lead to Late Departures**: If a previous flight arrives late, the aircraft has less time for ground operations, increasing the likelihood that the current flight will depart late.

2. **Operational Time Variability Compounds**: Variability in air time (weather, air traffic) and turnover time (gate availability, crew scheduling, maintenance) accumulates through the rotation, creating cascading delays.

3. **Network Effects Amplify Impact**: Delays at hub airports propagate through connecting flights, affecting multiple downstream flights in the network.

### Application to Feature Engineering

The primary motivation for our feature engineering strategy is to provide our models the information they need in order to estimate this equation. To capture cascading delay effects, we need information about previous flight operations. However, actual duration features contain data leakage (they use future information). Our multi-pronged approach includes:

1. **Safe Features**: Intelligent data leakage handling that uses actual values when available at prediction time, and falls back to scheduled times or estimates when not available.

2. **Graph Analysis Features**: PageRank-based network centrality metrics that capture airport congestivity and network effects, providing insights into delay propagation patterns.

3. **Meta-Models**: Train separate models to predict previous flight durations (air time, turnover time, rotation time) using only information available at prediction time, then use these predictions as features in our main departure delay model.


#### Flight Lineage Features

When an aircraft experiences delays on one flight, those delays cascade to subsequent flights in the same rotation. Flight lineage features capture the cascading effects of delays through aircraft rotations by including information in our dataset about the aircraft's previous rotation. 

##### Key Terminology

- **Air Time**: Time from wheels-off to wheels-on (actual flight duration, excluding all ground operations)
- **Turnover Time**: Time from previous flight's actual arrival to current flight's actual departure (includes taxi-in, gate operations, and taxi-out)
- **Rotation Time**: Time from previous flight's actual departure to current flight's actual departure (complete cycle = Air Time + Turnover Time)

These terms map directly to our fundamental theory of delay propagation, where **Delay = Available Time - Air Time - Turnover Time**.

![The Flight Lineage](https://raw.githubusercontent.com/Connor-Watson-Berkeley/flight-departure-delay-predictive-modeling/refs/heads/main/report-images/phase-3/Flight%20Lineage.png)

##### Concept: Aircraft Rotation Tracking

Flight lineage features connect each flight to its **previous flight in the same aircraft's rotation**. This enables our models to capture how delays propagate through an aircraft's daily schedule, directly supporting our fundamental theory of delay propagation.

**Implementation**: Previous flight data is retrieved using window functions (LAG) partitioned by tail number and ordered by arrival timestamp, ensuring we capture the correct sequence of flights for each aircraft.

##### Flight Lineage Feature Categories

###### 1. Core Sequence Features

These features capture the position and status of the current flight within the aircraft's rotation sequence:

- **Lineage Rank**: Rank of current flight in the aircraft's daily sequence (1 = first flight of the day)
- **Is Jump Flight**: Boolean flag indicating if the aircraft "jumped" airports (previous destination ≠ current origin)
  - **Business Context**: Jump flights represent aircraft repositioning, which may have different operational patterns
- **Is First Flight**: Boolean flag for first flight of the day
  - **Business Context**: First flights may have different patterns (overnight maintenance, early morning operations)
- **Number of Previous Flights**: Number of previous flights in the rotation

###### 2. Previous Rotation Attributes

We capture all same base feature families for the flight's previous rotation, providing comprehensive context about the aircraft's prior operations.

**Purpose**: These features enable models to understand the operational context of the previous flight, which directly impacts the current flight's readiness.

###### 3. Scheduled Time Features

These features reflect the airline's scheduled durations for the prior rotation, providing baseline expectations. They map directly to components of our fundamental equation: **Delay = Available Time - Air Time - Turnover Time**. We capture the carrier scheduled times for Rotation Time, Turnover Time, and Air Time based on scheduled arrival and departure times. These duration features represent the time the airline has budgeted for the trip.

###### 4. Safe Time Features (Data Leakage-Free)

Safe features intelligently handle data leakage by checking prediction cutoff timestamps (2 hours before scheduled departure). These features use actual values when available at prediction time, and apply conditional imputation logic when values are not realized before the cutoff window.

![The Flight Lineage](https://raw.githubusercontent.com/Connor-Watson-Berkeley/flight-departure-delay-predictive-modeling/refs/heads/main/report-images/phase-3/Safe%20Delay%20Features.png)

**Logic**:
- **If actual values were realized before the prediction window**: Use the actual value (it's known at prediction time)
- **If actual values were realized after the prediction window**: Impute based on scheduled time + assumptions (replicating what would be known in production)

**Safe Previous Arrival Delay**:
- If previous flight already arrived (actual_arr ≤ cutoff): use actual arrival delay
- If previous flight hasn't arrived yet (actual_arr > cutoff):
  - If scheduled_arr < cutoff: use (cutoff - scheduled_arr) in minutes (at least delayed until now)
  - If scheduled_arr ≥ cutoff: use 0 (scheduled in future, we assume no delay)

Safe Previous Departure Delay is calculated in a similar manner.

**Safe Rotation Time** → **AVAILABLE TIME (Safe)**:
- If previous flight already departed (actual_dep_time ≤ prediction_cutoff): use actual departure time
- If previous flight hasn't departed yet (actual_dep_time > prediction_cutoff):
  - If scheduled_dep_time ≥ prediction_cutoff (scheduled is in future): use scheduled departure time
  - If scheduled_dep_time < prediction_cutoff (scheduled is in past): use "right now" (prediction_cutoff)
- This provides the "Available Time" component of our fundamental equation in a data leakage-free manner, representing the time window available for the rotation at prediction time.

Safe air time, turnover time, and time since previous arrival features are calculated in a similar manner.

###### 5. Rotation Duration Actuals (Target Variables for Meta Models)

We compute actual rotation duration features that relate directly to our fundamental theory components:

- **Actual Rotation Time** → **AVAILABLE TIME (Actual)**: Actual time from previous flight's actual departure to current flight's actual departure
  - This represents the actual time window available for the rotation, but may not be known at prediction time.

- **Actual Turnover Time** → **TURNOVER TIME (Actual)**: Actual time from previous flight's actual arrival to current flight's actual departure
  - This represents actual ground operations time (taxi-in, gate operations, taxi-out), but may not be known at prediction time.

- **Previous Flight Actual Air Time** → **AIR TIME (Actual)**: Actual air time for previous flight (wheels-off to wheels-on)
  - This represents the actual flight duration, but may not be known at prediction time.

These features often or always (in the case of Rotation and Turnover time) contain data leakage so they **cannot be used directly in our models**. Instead, we trained **meta-models** to predict these duration components using only information available at prediction time (scheduled times, weather, airport characteristics), which we will discuss in the Meta Models section.

#### Graph Features

Graph features measure airport importance in the flight network using PageRank algorithms. Hub airports (high PageRank) are major connection points where delays have the greatest network-wide impact. These features help models identify which airports are most critical to delay propagation.

**Business Value**: Understanding network centrality enables models to prioritize delay mitigation at hub airports, where operational issues affect the most downstream flights. A delay at a major hub like Atlanta (ATL) cascades through many more flights than a delay at a smaller regional airport.

![Airport and Flight Graph](https://raw.githubusercontent.com/Connor-Watson-Berkeley/flight-departure-delay-predictive-modeling/refs/heads/main/report-images/phase-3/Airport%20%26%20Flight%20Graph.png)

#### Concept

PageRank metrics measure the importance of airports in the flight network by analyzing the structure of flight connections. Airports with high PageRank values are:
- **Hub airports**: Major connection points with many incoming and outgoing flights
- **High-traffic airports**: Experience more congestion and operational complexity
- **Network amplifiers**: Delays at these airports propagate more widely through the network

##### Implementation

PageRank is computed on the flight network graph where:
- **Nodes**: Airports
- **Edges**: Flight routes (origin → destination)
- **Edge weights** (for weighted PageRank): Route frequency (number of flights on that route)

The algorithm iteratively computes airport importance based on:
1. Number of incoming connections (flights arriving at the airport)
2. Importance of airports that connect to it (recursive importance)
3. Route frequency (for weighted version)

##### Weighted vs. Unweighted PageRank

- **Unweighted PageRank**: Treats all flight connections equally, measuring pure network structure
- **Weighted PageRank**: Weights connections by route frequency, capturing operational volume and importance

Both metrics provide complementary signals: unweighted captures structural importance, while weighted captures operational significance.

##### Feature Set

We compute PageRank metrics for both current and previous flight airports, organized by rotation, location, and method:

| Rotation | Location | Method | Description |
|----------|----------|--------|-------------|
| **Current Flight** | Origin | Weighted | Network importance of origin airport (weighted by route frequency) |
| **Current Flight** | Origin | Unweighted | Network importance of origin airport (unweighted) |
| **Current Flight** | Destination | Weighted | Network importance of destination airport (weighted by route frequency) |
| **Current Flight** | Destination | Unweighted | Network importance of destination airport (unweighted) |
| **Previous Flight** | Origin | Weighted | Network importance of previous flight's origin airport (weighted) |
| **Previous Flight** | Origin | Unweighted | Network importance of previous flight's origin airport (unweighted) |
| **Previous Flight** | Destination | Weighted | Network importance of previous flight's destination airport (weighted) |
| **Previous Flight** | Destination | Unweighted | Network importance of previous flight's destination airport (unweighted) |

**Total**: 8 PageRank features (2 rotations × 2 locations × 2 methods)

**Why We Include Both Previous Flight Destination and Current Flight Origin**: In normal aircraft rotations, the previous flight's destination airport and the current flight's origin airport are the same (the aircraft lands and then departs from the same airport). However, they can differ in edge cases such as data loss or aircraft relocation ("jump flights"). We compute PageRank for both because:
- **In meta-models**: We use the previous flight's destination airport PageRank for air time predictions to avoid potential errors from edge cases where the airports differ

##### Impact on Flight Delays

Graph features support our fundamental equation for delay propagation, specifically the **Network Effects Amplify Impact** mechanism:

- **Hub airports** (high PageRank) have high flight density and limited gates, creating bottlenecks
- Delays at hub airports affect multiple aircraft rotations simultaneously
- Gate congestion at hubs creates cascading delays across multiple flights
- Network centrality metrics help models identify which airports are most critical to network-wide delay propagation

##### Current Limitation: Missing Temporal Interactions

While graph features capture network congestivity (which airports are hubs), we **did not fully capitalize on the interactions** between network centrality and cascading delay effects.

**Missing Feature**: Windowed rolling aggregates of delays at airports before the prediction window.

**Rationale**: This would allow models to learn the interaction between:
- **Network congestivity** (PageRank): Which airports are hubs
- **Current operational status** (rolling delays): What's happening at those hubs right now

**Example**: A high-PageRank hub airport (e.g., ATL) with high recent delays (4-hour rolling average) would signal much stronger cascading delay risk than a high-PageRank hub with normal operations.

# Model Pipeline

## Data Splitter and Cross Validator
While specific models have slightly different pipelines to suit the requirements of the models, the main model pipeline can be characterized to have 3 phases:

1. Data Splitting: A **Data Splitter** splits the 3-month, 12-month, and 60-month data respectively based on the flight date and checkpoints them into shared storage. Specifically, each dataset is divided into 5 equal portions sequentially for cross validation. Engineered features are subsiquently applied to each split.
2. Data Loading: A **Data Loader** loads the previously checkpointed data into memory for consumption of the cross validator.
3. Cross Validator: A **Cross Validator** uses time-series rolling-window cross validation to train model on 3 rolling folds. A rolling-window approach is used instead of expanding-window to improve scalability: compared to expanding window approach, which uses all prior data in the training, rolling-window only uses the most recent data within the window. Because the data is pre-split and checkpointed, the pipeline avoids re-computation each time.

We designed the pipeline such that feature engineering is performed at the cross validation stage. This is intentional, as some models prefer different feature engineering approaches, and one uniform feature engineering pipeline would not fit all. For example, after loading the data, the pipeline would engineer the graph feature based on the existing features, and then derive meta model features based on that.

Since we engineer time-series features based on information prior to each flight, this approach also ensures minimal data leakage: when the cross validator fits a model, it only has access to the training fold and must perform all feature engineering based on it only.

The design of the pipeline practices object-oriented programming: data splitter, data loader, and cross validator are implemented as separate modules with interfaces. This ensures that each stage can be separately customized and updated, streamlining collaboration.

## Data Leakage & Generalization

Data leakage occurs when a model uses information that would not be available in a production setting, leading to overly optimistic performance estimates that don't reflect real-world accuracy. Our pipeline rigorously prevents data leakage through temporal cross-validation, safe feature engineering, and careful separation of training and inference for meta-models and graph features. This ensures our model performance metrics accurately reflect what can be achieved in production when predicting departure delays 2 hours before scheduled departure.

### What is Data Leakage?

**Data leakage** occurs when a machine learning model uses information during training that would not be available at prediction time in a production environment. This creates an unrealistic advantage that inflates model performance metrics, making the model appear more accurate than it actually would be when deployed.

### Why Data Leakage Matters: Business Impact

Data leakage creates **false confidence** in model performance. A poor performing model that appers performant due to data leakage is far more devastating compared a poorly performang model that appears poor. The former may be deployed and trusted, while the latter is caught before deployment. When a model with data leakage is deployed, it fails to meet expectations, leading to:

- **Invalid Policy Decisions**: Strategic decisions based on inflated performance metrics become unreliable, potentially leading to poor resource allocation and operational planning
- **Operational Disruption**: Airlines trusting unreliable predictions may make suboptimal decisions about gate assignments, crew scheduling, and passenger communications
- **Loss of Trust**: When models underperform in production, stakeholders lose confidence in the predictive system and the teams which produced them, hindering adoption and future improvements

By rigorously preventing data leakage, we ensure that performance metrics reflect true production capabilities, enabling reliable decision-making and operational planning.

### Cardinal Sins of Machine Learning

We ensure our pipeline does not violate any cardinal sins of ML:

**No Target Leakage**
- We never use the target variable (`dep_delay`) or any direct derivative of it as a feature
- Meta-models predict **components** of the delay equation (air time, turnover time), not the delay itself

**No Train-Test Contamination**
- Temporal cross-validation ensures strict temporal separation
- Graph features and meta-models are trained only on training folds
- No information from validation or test sets leaks into training

**No Future Information**
- All features are evaluated at the prediction cutoff (2 hours before scheduled departure)
- Safe features explicitly check prediction cutoff timestamps
- Scheduled times are used when actual times are unavailable

**No Data Snooping**
- Model selection and hyperparameter tuning are performed using only training and validation sets
- Test set is used only for final performance evaluation
- No iterative refinement based on test set performance

**No Preprocessing Leakage**
- All preprocessing (imputation, scaling, encoding) is fit only on training data
- Statistics (means, standard deviations, encodings) are computed from training folds only
- Validation and test sets are transformed using training-set statistics


### Data Leakage in Flight Delay Prediction

Flight delay prediction presents unique data leakage challenges due to **cascading delay effects** and **temporal dependencies**. When an aircraft experiences delays on one flight, those delays propagate to subsequent flights in the same rotation. These temporal dependencies would leak information from training data into test sets if not appropriately accounted for.

##### Temporal Dependence Leakage Risk

As we've demonstrated, flight delays are not independent events due to the cascading effect of delays. This temporal dependence creates a critical data leakage risk when using random train/test splits.

**The Problem with Random Splits**:

With a random split on flight data, there is no temporal ordering, flights can be randomly assigned to training or test sets regardless of when they occurred. This creates data leakage because flights at the exact same time (or very close times) can be split between train and test:

- **Training set**: Includes some EWR flights from March 5th, 2018 at 2:00 PM
- **Test set**: Includes other EWR flights from March 5th, 2018 at 2:00 PM

The model learns from some flights at 2:00 PM and uses that information to predict delays for other flights at the same time. This directly leaks information about conditions at that specific moment, creating an unrealistic performance estimate that won't hold in production, where we must predict delays without knowledge of how other flights at the same time actually performed.

**Why This Matters**:

This is analogous to the human cognitive bias known as "hindsight bias" or "post-hoc reasoning," where after an event occurs, we may say "I knew that would happen, there was so much evidence the event would occur." However, in advance of the event occurring, we would not have felt the same confidence in making the same predictions. Similarly, a model trained on future data appears more confident and accurate than it actually would be when making predictions in real-time.

###### Flight Lineage Leakage Risk
Delays cascade through aircraft rotations:
1. Flight A arrives late → reduces available time for ground operations
2. Flight B (same aircraft) departs late → reduces available time for Flight B's operations
3. Flight C (same aircraft) is affected by Flight B's delay

If we use actual delay information from Flight B that was not realized two hours in advance when predicting Flight C's delay, we're using information that wouldn't be available in production at prediction time. As we will discuss later, these cascading delays are of critical importance for our use case, so we must find techniques to learn from this information while ensuring our models can perform when this information is not available

### Techniques to Prevent Data Leakage

We implemented multiple techniques to eliminate data leakage risks, ensuring our models only use information that would be available in production. While these techniques are discussed in detail in other sections, we emphasize their data leakage prevention aspects here.

#### 1. Temporal Cross-Validation

**Data Leakage Prevention**: Temporal cross-validation with rolling window splits ensures strict temporal separation, training data always comes from time periods **before** validation/test data, preventing any future information from influencing past predictions.

**Implementation**: (See the main model pipeline section for details.) This approach directly addresses temporal dependence leakage by respecting the temporal sequence of events, ensuring models learn from prior events to predict future events.

#### 2. Safe Lineage Features

**Data Leakage Prevention**: For features that depend on previous flight information, we implemented **"Safe Lineage Features"** that intelligently handle data leakage by checking prediction cutoff timestamps:

- **If actual values were realized before the prediction window**: Use the actual value (it's known at prediction time)
- **If actual values were realized after the prediction window**: Impute based on scheduled time + assumptions (replicating what would be known in production)

**Examples**: `safe_prev_departure_delay`, `safe_lineage_rotation_time_minutes`, `safe_time_since_prev_arrival` (See the flight lineage features section for detailed implementation.)

#### 3. Meta-Models for Unrealized Information

**Data Leakage Prevention**: For information that **cannot be realized until after the current flight's departure**, the thing we're trying to predict, we implemented **meta-models** that predict these values using only information available at prediction time.

**Implementation**: Meta-models predict previous flight air time, turnover time, and rotation time using scheduled times, weather, and airport characteristics, all available at prediction time. These predictions are then used as features in our main departure delay model. (See meta-model feature engineering section for detailed implementation and performance results.)

#### 4. Strict Training/Inference Separation for Graph Features and Meta-Models

**Data Leakage Prevention**: To ensure graph features and meta-models did not train on information from test sets, we implemented strict separation:

- **Graph features**: Computed from network structure using only training data
- **Meta-models**: Trained **only on training folds**, then inferred on cross-validation and test folds
- **Held-out sets**: Never used for training graph features or meta-models

This ensures true generalization, our main models learn to use predictions from models that were trained on different data, replicating the production scenario.

**Future Improvement**: For even more robust generalization, we could train graph features and meta-models on a **prior** split set for the training sets as well. While the current implimentation does not cause data leakage (cross-validation and test sets are still respected), it could improve model performance as models would learn how to interact with predictions from models trained on past data. However, the strong generalization demonstrated in our meta-models section indicates that meta-model predictions perform similarly on unseen data, so the performance gain may be minor.

#### 5. Conservative Feature Exclusion

**Data Leakage Prevention**: Low-risk cumulative delay features that depend on rotations **before** the prior rotation were tested but **omitted from final reported models**. While extremely unlikely to contain data leakage (it would require an airplane to make two full rotations in under 2 hours), it is still technically possible and not explicitly controlled for.

**Future Recommendation**: Impliment temporal exclusion rather than lineage sequence exclusion to maintain cumulative and rolling average delays, while adding explicit logic to exclude any values which were not realized before our prediction window.

### Pipeline Data Leakage Prevention Summary

Our pipeline prevents data leakage through multiple layers of protection:

1. **Temporal Structure**: Rolling window cross-validation ensures temporal ordering, training data always precedes validation/test data
2. **Feature Engineering**: Safe features check prediction cutoff timestamps; meta-models predict unrealized information using only available data
3. **Model Training Separation**: Graph features and meta-models are trained only on training folds, inferred on held-out sets
4. **Explicit Exclusion**: Actual time features and cumulative features are excluded if they cannot be guaranteed to exclude unrealized values

These rigorous protections have allowed our model to learn from highly predictive information while ensuring performance will be maintained in production enviroments by simulating the real-world avaiability of this information at prediction time in every step of our model development process.

Our model is a regression model designed to predict flight delays (in minutes). To evaluate its performance, we use a combination of standard regression metrics and domain-specific operational metrics. ###Loss Function:
####  Mean Squared Error (MSE)
Our models predict continuous flight delay times in minutes using Mean Squared Error (MSE) as its loss function during training, which penalizes large prediction errors more heavily than small ones. It is useful as our loss function as it measures the average squared difference between predicted and actual delay times. MSE penalizes large errors more than small errors due to the squaring operation, making it sensitive to outliers and ensuring the model focuses on reducing substantial prediction errors. MSE helps us build delay models that prioritize reducing large, operationally costly timing errors (e.g., missing connections, crew misalignments) by heavily penalizing big misses in predicted departure delay minutes.

Mean Squared Error (MSE), calculated as 

$$MSE = \frac{1}{n}\sum_{i=1}^{n}(y_i - \hat{y}i)^2$$

| Loss Function | Formula | Description |
|--------|---------|-------------|
| Mean Squared Error (MSE) | $$\mathrm{MSE} = \frac{1}{N} \sum_{i=1}^{N} (y_i - \hat{y}_i)^2$$ | Mean Squared Error - average of squared prediction errors, useful for optimization but less interpretable than RMSE |

###Key Regression Metrics
#### Root Mean Squared Error (RMSE)
We've selected RMSE (the square root of MSE) as one of our key regression metrics as it is expressed in the same units as the target variable (minutes), providing a more interpretable measure of average prediction error magnitude than MSE which can be directly understood by stakeholders. RMSE gives operations and planning teams an easily interpretable estimate of typical delay error in minutes, so they can judge whether the model’s predictions are accurate enough to reliably support gate planning, crew scheduling, and connection risk decisions.

The Root Mean Squared Error (RMSE), computed as

$$\mathrm{RMSE} = \sqrt{\frac{1}{N} \sum_{i=1}^{N} (y_i - \hat{y}_i)^2}$$

| Metric | Formula | Description |
|--------|---------|-------------|
| Root Mean Squared Error (RMSE) | $$\mathrm{RMSE} = \sqrt{\frac{1}{N} \sum_{i=1}^{N} (y_i - \hat{y}_i)^2}$$ | Root Mean Squared Error - measures average prediction error magnitude in original units, penalizes large errors more heavily |

#### Mean Absolute Error (MAE)

Additonally, we assess our models on Mean Absolute Error (MAE), defined as

$$
MAE = \frac{1}{n}\\sum_{i=1}^{N}|y_i - \hat{y}i|
$$

calculates the average absolute difference between predicted and actual delays without squaring, making it less sensitive to outliers than MSE or RMSE and providing a robust measure of typical prediction error. MAE gives us a robust, easy‑to‑interpret view of typical delay error in minutes, less skewed by rare extreme disruptions so stakeholders can understand how far off predictions are for the vast majority of everyday flights.

| Metric | Formula | Description |
|--------|---------|-------------|
| Mean Absolute Error (MAE) | see above | Mean absolute difference between predicted and actual values. |


###Key Classification Metrics
Beyond standard regression metrics, we evaluate operational utility through domain-specific classification metrics to help business stakeholders understand our models prediction outcomes.

#### On-Time Prediction Accuracy (OTPA)
OTPA (On-Time Prediction Accuracy) measures binary classification accuracy for the critical 15-minute on-time threshold defined by the U.S. Department of Transportation.

**On-Time Prediction Accuracy** measures the accuracy of predicting flights within a narrow tolerance window (+/- 5 minutes), which is critical for operational planning because resource allocation decisions, such as gate assignments, crew scheduling, and passenger communication, require precise delay estimates rather than broad categorizations. 

OTPA converts delay predictions into a binary outcome: *on-time* (delay < 15 minutes) or *delayed* (≥ 15 minutes), and measures accuracy or F1-score. This bridges regression and classification perspectives, aligning with how the aviation industry defines on-time performance. On-Time flights help improve the reputation of the airline and keep the airport traffic smoothly moving. 

We operationalize this metric by converting continuous delay predictions into binary outcomes consistent with U.S. Department of Transportation (DOT) definitions:

- On-Time: delay < 15 minutes
- Delayed: delay ≥ 15 minutes

**On-Time Definition:**

- Predicted On-Time = 1 if `ŷ < 15`, else 0  
- Actual On-Time = 1 if `y < 15`, else 0  

**Formula:**

`On-Time Prediction Accuracy (OTPA) = (TP + TN) / (TP + TN + FP + FN)`

Where:  
- **TP** = correctly predicted on-time flights  
- **TN** = correctly predicted delayed flights  
- **FP** = predicted on-time but actually delayed  
- **FN** = predicted delayed but actually on-time  

| Metric | Formula | Description |
|--------|---------|-------------|
| **On Time Prediction Accuracy (OTPA)** | $$\text{OTPA} = \frac{TP + TN}{TP + TN + FP + FN}$$ | Binary classification accuracy for on-time (<15 min) vs delayed (≥15 min) flights |

This metric translates regression outputs into actionable insights for **crew scheduling**, **resource allocation**, and **passenger notifications**, emphasizing precision in predicting near-schedule flights.

#### Severe Delay Detection Rate (SDDR)
Severe Delay Detection Rate measures the recall specifically for delays >=60 minutes, which is important for identifying flights requiring major operational adjustments such as aircraft substitutions, crew reassignments, or passenger compensation, as these severe delays have the most significant impact on operations and customer satisfaction. 

Buckets:
- OnTime: < 15 minutes
- Delayed: 15-60 minutes
- Severe: 60+ minutes

Formula:

Severe Delay Detection Rate (SDDR) = TP_severe / (TP_severe + FN_severe)

Where:

- **TP_severe** = number of flights correctly predicted to have severe delays (≥ 60 min)
- **FN_severe** = number of severely delayed flights the model failed to identify

| Metric | Formula | Description |
|--------|---------|-------------|
| **Severe Delay Detection Recall (SDDR)** | $$\text{SDDR} = \frac{TP}{TP + FN}$$ | Proportion of severe delays (≥60 miinutes) correctly identified - critical for operational planning |

A high SDDR ensures the model can proactively flag flights requiring contingency actions such as aircraft swaps, gate rescheduling, and passenger rebooking. This metric operationalizes the model’s real-world value, not just statistical accuracy, but its usefulness for delay mitigation and resource optimization.

### Loss Function and Regularization

The **loss function** used for both Gradient Boosted Trees and XGBoost models is **mean squared error**, described in the metrics section above.

The combined loss and L2 ridge regularization function is:

$$ loss = MSE + \lambda*\sum(w_i^2) $$

where MSE is the mean squared error, lambda is the regularization parameter, and w_i represents the model coefficients.

## Neural Network Implementation & Experimentation

To address the complexity of the multi million row dataset, we moved beyond standard Spark MLlib classifiers and implemented a custom Deep Learning solution using **PyTorch**. This approach allowed us to leverage GPU acceleration and distributed training, which are essential for training high-capacity models on such a large volume of data.

## Neural Network Architecture (MLP)

We implemented a **Multilayer Perceptron (MLP) Regressor** designed to predict flight departure delays (in minutes). The architecture was built using `torch.nn` and is fully dynamic, allowing for variable depth and width during experimentation.

### **Model Structure**
The network consists of a sequence of fully connected blocks. For a defined list of hidden units (e.g., `[512, 256, 128]`), the architecture follows this pattern for each layer:

1.  **Linear Layer:** Projects input dimension to hidden dimension.
2.  **Batch Normalization (BatchNorm1d):** Normalizes layer inputs to stabilize training and accelerate convergence.
3.  **Activation (ReLU):** Introduces non-linearity.
4.  **Dropout (p=0.1 - 0.3):** Regularization to prevent overfitting on the training shards.

The network is structured as a sequence of fully connected blocks, where for each hidden layer _l_:

$$\text{Output}_l = \text{Dropout}(\text{ReLU}(\text{BatchNorm}(\text{Linear}(\text{Input}_{l-1}))))$$

The final layer is a single linear unit, outputting the predicted delay scalar.


## Distributed Training Implementation 

To satisfy the requirement for distributed training on clusters, we implemented a custom estimator **`SparkPyTorchEstimator`** that utilizes Databricks' **`TorchDistributor`**. This allows us to train the PyTorch model in a data-parallel fashion across multiple GPU-enabled worker nodes.


### **Key Technical Features:**
* **Data Parallelism (DDP):** We utilized `torch.nn.parallel.DistributedDataParallel` to synchronize gradients across workers.
* **Custom Iterable Dataset:** Instead of collecting data to the driver (which causes OOM errors with 22M rows), we implemented a `ParquetFlightIterableDataset`. This custom class allows workers to stream data directly from distributed Parquet shards on DBFS during the training loop.
* **Sharding Strategy:** The 22M row dataset was repartitioned into ~8,000 shards to ensure manageable file sizes for the streaming dataloader.
* **Early Stopping:** A custom `EarlyStopping` class was implemented to halt training if validation loss did not improve for a set `patience` (e.g., 7 epochs), saving computational resources.

We constructed a unified pipeline that integrates Spark-based feature engineering with the deep learning model. Please refer to the modeling pipeline section and the feature engineering sections above for details.

## Novel Approach - Multi-Head Multi-Task Model (MLP + GBT + Logistic Regression)
In response to the fundamental challenges faced in our earlier model iterations posed by our highly skewed delay distribution and competing evaluation metrics, we experimented with a multi-head, end-to-end modeling pipeline that jointly optimizes for on-time performance accuracy (OTPA), regression RMSE, and severe delay detection rate (SDDR). Our final tuned Neural Networks and Tree Based Ensemble models ended up performing well on these metrics but we did explore having different specialized models optimized for each metric in the event that our competing metrics couldn't be properly handled by a single model.

Early iterations of our single-head models were challenged in this setting: multi-class bucket models (on-time/delayed/severely delayed) faced a hard tradeoff between recall and accuracy, and when we instead regressed on continuous delay and post-bucketed predictions, the inherently high RMSE on this noisy problem propagated directly into poor bucket assignments. Later models' improved performance on RMSE greatly impacted the performance of OTPA and SDDR making the need for the multi-head approach unnecessary but did prove useful in exploring some of the fundamental assumptions about our data and its skewed distribution. In particular, optimizing aggressively for SDDR on a highly imbalanced label (severe delays are a tiny fraction of flights) tended to collapse OTPA, whereas optimizing for OTPA or RMSE often caused the model to miss almost all severe events. To address this, we implemented a multi-head model that shares a single, rich feature-preprocessing pipeline but trains three heads simultaneously, each aligned with one key metric.

#### Multi-Head Pipeline
The shared pipeline begins by loading pre-computed graph-based features and lineage-based features from the cross validation folds, then performs systematic cleaning and transformation. We also engineered tiered indicator flags to better capture non-linear structure and missingness: for example, visibility_missing_flag, very_bad_vis_flag, and bad_vis_flag detect missing or extremely poor visibility. Crucially, we plugged this multi-head estimator into a modified cross-validation wrapper that can consume and evaluate multiple prediction columns per fold. Our MultiHeadCVFinal class wraps the existing cv.py infrastructure but, instead of a single prediction column, tracks three separate outputs and computes head-specific metrics per fold. For each CV fold, we fit the shared-preprocessing plus three-head model on the training split, then generate predictions on both train and validation splits. A custom evaluator aggregates per-head performance into a pandas DataFrame, returning per-fold Train/Val rows for nn_otpa, reg_rmse, severe_sddr, and severe_sddr_prec, along with mean and standard deviation summaries.

#### Alternative Models Explored
On top of this shared representation, we train three separate models per fold exploring different models and their possible effectiveness on their targeted metric. We explored using a Binary Categorical Neural Network for OTPA, an MLlib GBTRegressor for RMSE and a Logistic Regresion classifier for SDDR and Severe Delay Detection Precision. These different model explorations helped solidify that our selected models were the most effective for the problem at hand.

This chart shows us that **our model systematically underpredicts the magnitude of delays**, particularly for severe delays. For small delays (such as between 1-15 minutes), predictions are fairly close to actuals. For mid-range delays (60-120 minutes), predictions are roughly 40-50% of actual values. With extreme delays of duration of 180 or more minutes, the actual mean delay is ~320 minutes, while the predicted is only ~70 minutes. As actual delays grow non-linearly, while predicted delays grow almost linearly and much more slowly, this suggests the model struggles to capture extreme events, and instead regresses predictions toward the mean.

What this means practically is that the model **may be acceptable for short-delay forecasting**, or **ranking or categorizing delays**; however, it would be risky for operational planning for major disruptions or customer-facing estimates during severe delays.

### Possible causes for differences

One possible reason for the differences between predicted and actual delay times is **our loss function of mean squared error penalizes large errors heavily**. We could investigate a different loss function that does not penalize the large errors as much.

We may also **lack insufficient examples of extreme delays** in our training data - we did show earlier that the dataset is highly imbalanced, with around 81% of the data showing on-time flights.

Another reason may be that our **current features don't capture large disruption drivers** such as severe weather or other extreme incidents.

## Model drift over time

We did notice a significant increase in RMSE/MAE as we moved from training on our 3 month dataset to the 1-year (12M) and 5-year (60M) datasets. For example, here's a table of the test RMSE for the same set of features over different time spans:

| Model | Features | Time Span (Months) | Test RMSE |
|-------|----------|--------------------|-----------|
| XGBoost | Graph, Lineage | 3 | 31.34 |
| XGBoost | Graph, Lineage | 12 | 33.76 |
| XGBoost | Graph, Lineage | 60 | 45.46 |

One can see the significant jump in RMSE as the time span increases by 5x (which is an RMSE increase of around 45%). One way to mitigate this could be with **exponential smoothing**. This technique would give greater weight to the most recent observations (in this case, the observations closest to 2019) and would reduce observations exponentially as the distance from the observations rises.

Practically, one way we could implement this is by generating predictions from our 3 training models - one trained on the 2015-2016 data, another model trained on the 2016-2017 data, and another trained on the 2017-2018 data - then, weighting the predictions using exponential smoothing and summing them up together. In this way we are creating a sort of ensemble model where the predictions are weighted using exponential smoothing. 



## Engineered Features & Data Completeness Drive Substantial Performance Improvements

The biggest performance improvements came not from how we fine-tuned our models, but rather from how we thought about our data. Progressive addition of engineered features demonstrates how different feature families address distinct aspects of delay prediction. Each layer captures complementary signals: lineage features track cascading delays through aircraft rotations, graph features capture network-wide congestion effects, and meta-models predict operational constraints.

**XGBoost, 60M dataset, 2019 test set**:

| Additional Features | Source | RMSE (min) | OTPA (%) | SDDR (%) | Improvement |
|---------------------|---------|------------|----------|----------|-------------|
| *(Base Features Only)* | Custom | 47.63 | 74.77% | 1.06% | Baseline |
| Graph + Lineage | Custom | 45.46 | 79.74% | 28.82% | +5.0% OTPA, +27.8% SDDR |
| Graph + Lineage + Meta-Models | Custom | 43.13 | 82.08% | 39.93% | +7.3% OTPA, +38.9% SDDR |
| Graph + Lineage + Meta-Models | OTPW | 37.76 | 87.47% | 62.94% | **+12.7% OTPA, +61.9% SDDR** |

**What makes these features work?** The answer lies in understanding temporal dependencies. Flight delays don't occur in isolation, they cascade through aircraft rotations. When we model a flight's delay, the most recent rotation is highly predictive, similar to how attention mechanisms in sequence models focus on the most relevant context. 

However, when models require temporal dependencies like the most recent flight in the sequence, missingness is detrimental to model performance. Missing flights break the lineage chain, forcing the model to use incorrect or outdated information (e.g., using a flight from two rotations ago when the immediate previous flight is missing). This introduces noise and reduces the model's ability to identify true cascading delay effects. Our results demonstrate this clearly: the OTPW dataset (complete data) achieved +5.4% OTPA and +23.0% SDDR improvement over the Custom dataset (data loss), showing that complete temporal sequences are essential for models that rely on lineage relationships.

This insight suggests future work could leverage transformer-based architectures. Instead of simply modeling the current and previous rotation as separate features, a transformer could train directly on the sequence of flights for each tail number. This would allow the model to learn the interconnectedness of cascading delay effects across an aircraft's entire rotation sequence, with attention mechanisms naturally identifying which previous flights are most relevant for predicting the current delay. Maintenance events could be embedded in the sequence, where attention mechanisms can identify long gaps between maintenance and recognize increasing risk of substantial delays.

The improvements we see 27.8% SDDR from graph + lineage, +11.1% SDDR from meta-models, demonstrate that capturing temporal dependencies is critical. Lineage features capture aircraft-specific propagation, graph features address network-wide congestion effects, and meta-models predict operational constraints (air time, turnover time) that directly impact delay severity. Together, they validate our fundamental equation **Delay = Available Time - Air Time - Turnover Time** as a framework for understanding delay mechanisms.

**Irreducible Error and the Dataset Ceiling**

This finding connects to the concept of **irreducible error** in the bias-variance decomposition. The expected prediction error can be decomposed as:

$$E[(y - \hat{y})^2] = \text{Bias}^2 + \text{Variance} + \text{Irreducible Error}$$

A common misconception is that irreducible error represents an absolute ceiling on model performance. However, irreducible error really describes the best a model can perform *given a particular dataset*. The perceived "noise" in your data can often be information that hasn't been captured yet.

There's an important irony here: **all the engineered features we created were already in the dataset**, they just weren't directly available to the model until we engineered them. Consider predicting a flight's delay based on weather, departure airport, and time of day. The error in predictions might appear as random noise. However, once we include the information that "the previous flight takes 3 hours and hasn't departed yet" (available 2 hours in advance), suddenly there is certainty that the current flight will be delayed. This information was always present in the data, we just needed to join flights by tail number and compute the lineage relationships.

This insight also lends itself to transformer-based architectures. Traditional models train on single rows at a time, requiring us to manually engineer features that capture relationships across rows. Transformers, however, can train on training examples that span a full lineage sequence, allowing the model to learn these relationships directly from the data structure itself. By presenting the model with sequences of flights for each aircraft, rather than isolated flight records, we enable the model to discover temporal dependencies that we would otherwise need to engineer manually.

**Irreducible error is the ceiling of your dataset, not an absolute ceiling, and can be addressed by thinking about your data differently.** Our 20.7% RMSE improvement demonstrates that what appeared to be irreducible error was actually reducible through capturing temporal dependencies and operational constraints that were present in the data but not initially accessible to the model.

## Turnover Time (Ground Operations) Causes the Greatest Uncertainty

Our meta-models found that predicting turnover time, the time between wheels on and wheels off, is substantially harder to predict than air time (10x larger RMSE).

| Target | Split | RMSE (min) | MAE (min) | MAPE (%) |
|--------|-------|------------|-----------|----------|
| **Previous Flight Air Time** | TEST | 14.50 | 7.42 | 80.3 |
| **Previous Flight Turnover Time** | TEST | 149.64 | 47.22 | 94.1 |

Accurately predicting turnover times will be key for further improvements of our models. Based on the MAE, we can predict air time within an average of ±7.42 minutes, which substantially differs from our ability to predict turnover time (±47.22 minutes). This provides two key insights: (1) the features in our dataset are more predictive for air time, and (2) turnover time is simply harder to predict. While air time is bounded by available fuel and distance, the unbounded nature of turnover time delays, which could be minutes, hours, or even days in the event of a repair or relocation, makes these predictions inherently harder. More consideration needs to go into modeling the skew of the turnover time, and more predictive features related to the causes of long turnovers will also improve our ability to predict these events.

**For modeling departure delays**: More feature development to capture variance in ground operations (scheduled or ad-hoc maintenance events, pre-flight actions, crew rotations, gate availability, baggage handling, fueling operations, etc.) will have a greater impact than improving air time predictions. Air time's low RMSE of 14.50, substantially lower than our best departure delay model's RMSE of 37.76, indicates the majority of variance is already captured, further improvements will yield diminishing returns. Turnover time's high RMSE of 149.64 suggests substantial room for improvement through better feature engineering.

**For airline operators**: Investment into more consistent and predictable turnover times will yield fewer delays. The high variance in turnover time (149.64 min RMSE) directly translates to delay uncertainty, as delays cascade through aircraft rotations when ground operations take longer than expected. 

To enable better predictive modeling, rigorous logging and public publishing of ground operations events is critical. This includes equipment availability (e.g., wing defrosters at northern airports), maintenance logs, baggage handling times, boarding and deplaning times, gate queue times, air traffic control bandwidth, and other operational factors. Making these datasets publicly available would enable the broader data science community to enhance predictive models and drive industry-wide improvements. Implementation of nation-wide logging policies for data completeness would ensure comprehensive coverage of ground operations events across all airports and carriers.

# Conclusion

This project successfully developed a machine learning pipeline to predict flight departure delays two hours before scheduled departure. Our central hypothesis was that engineered features capturing cascading delay effects through aircraft rotations, network dynamics, and weather patterns would substantially outperform baseline approaches.

We validated this hypothesis through four key feature engineering approaches:

1. **Flight Lineage Features** tracking cascading delays through aircraft rotations with safe temporal handling to prevent data leakage;
2. **Graph-based PageRank Features** identifying hub airport congestion patterns;
3. **Meta-Models** predicting previous flight durations using only information available at prediction time;
4. **Comprehensive Weather Integration** with proper temporal alignment. We implemented and compared Gradient Boosted Trees, XGBoost, and deep neural networks with distributed PyTorch training.

We demonstrated that, without data leakage, we can predict departure delays with an error of ±12.79 minutes on an unseen 2019 test set. Our best-performing model, XGBoost with Graph, Lineage, and Meta-Model features on the complete OTPW dataset, achieved 37.76 minutes RMSE (21% improvement over baseline), 12.79 minutes MAE (37% improvement), 87.47% OTPA, and 62.94% SDDR F1. We found that engineered features provide substantial value (8.4% RMSE reduction from lineage features alone), data quality is paramount (complete dataset enabled 23% SDDR improvement), and ground operations (turnover time) represent the primary source of delay uncertainty and operational leverage point for airlines.

For future work, we would consider temporal interaction features combining network centrality with rolling delay windows, enhanced cumulative features with explicit temporal exclusion, improved ground operations modeling focusing on gate availability and crew scheduling, sequence models (LSTMs/Transformers) for rotation dependencies, and production deployment with real-time streaming updates and model monitoring.

Overall, this project demonstrates that rigorous feature engineering combined with advanced modeling and careful data leakage prevention can meaningfully improve flight delay prediction while maintaining production viability. The framework provides actionable insights for both model development and operational strategy in an industry where minutes matter.

## Meta Model Deep Dive
### Model Selection

We considered several modeling approaches to prove the meta-model strategy:

1. **Linear Regression**: Too simple (high bias) - would not capture the complex non-linear relationships between features and flight durations.

2. **Neural Network**: Would likely take too long to run/evaluate and require significant effort to tune (high variance risk). Given this was a proof-of-concept effort, the tuning overhead was prohibitive.

3. **Random Forest**: Offers a good balance of bias and variance, with strong predictive performance out-of-the-box requiring very minimal hyperparameter tuning. This was our initial choice.

4. **XGBoost**: Selected after benchmarking showed similar performance to Random Forest but with dramatically faster training times.

## Regression Metrics
| Metric | Formula | Description |
|--------|---------|-------------|
| RMSE | $$\mathrm{RMSE} = \sqrt{\frac{1}{N} \sum_{i=1}^{N} (y_i - \hat{y}_i)^2}$$ | Root Mean Squared Error - measures average prediction error magnitude in original units, penalizes large errors more heavily |
| MAE | $$\mathrm{MAE} = \frac{1}{N} \sum_{i=1}^{N} \|y_i - \hat{y}_i\|$$ | Mean Absolute Error - measures average absolute prediction error, more robust to outliers than RMSE |
| R² | $$R^2 = 1 - \frac{\sum_{i=1}^{N} (y_i - \hat{y}_i)^2}{\sum_{i=1}^{N} (y_i - \bar{y})^2}$$ | Coefficient of Determination - proportion of variance in the target variable explained by the model (0-1 scale) |
| MSE | $$\mathrm{MSE} = \frac{1}{N} \sum_{i=1}^{N} (y_i - \hat{y}_i)^2$$ | Mean Squared Error - average of squared prediction errors, useful for optimization but less interpretable than RMSE |

## Classification Metrics
| Metric | Formula | Description |
|--------|---------|-------------|
| **OTPA Accuracy** | $$\text{OTPA} = \frac{TP + TN}{TP + TN + FP + FN}$$ | Binary classification accuracy for on-time (<15 min) vs delayed (≥15 min) flights |
| **OTPA Recall** | $$\text{Recall} = \frac{TP}{TP + FN}$$ | Proportion of actual on-time flights correctly identified |
| **SDDR (Recall)** | $$\text{SDDR} = \frac{TP_{severe}}{TP_{severe} + FN_{severe}}$$ | Proportion of severe delays (≥60 min) correctly identified - critical for operational planning |
| **Bucket Accuracy** | $$\text{Accuracy} = \frac{\text{Correct Predictions}}{\text{Total Predictions}}$$ | Overall accuracy across all 4 delay buckets (Early, OnTime, Delayed, Severe) |
